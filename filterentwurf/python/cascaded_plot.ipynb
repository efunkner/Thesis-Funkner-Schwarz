```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import freqz
%matplotlib tk
```

```python
# Hier die Koeffizienten des Filters einfügen
b = [0.00010725, 0.0, -0.00010725]
a = [1.0, -1.97944516, 0.99978549]
fs = 44100

# Frequenzgang berechnen
w, h = freqz(b, a, worN = 1024, fs = fs)

# Gesamter Freqeunzgang (Filterkaskadierung anhand der Potenz)
h_total = h**2
```

```python
# Plot
plt.figure(figsize= (12,4))
plt.plot(w, 20 * np.log10(np.maximum(np.abs(h_total),1e-10)), label='Kaskadierung')
plt.plot(w, 20 * np.log10(np.maximum(np.abs(h),1e-10)),'--', label='Grundverlauf')
plt.title('Frequenzgang: Filterkaskadierung')
plt.grid()
plt.xlim([0, 20000])
plt.ylim([-80, 10])
plt.legend()


plt.tight_layout()
plt.show()
```

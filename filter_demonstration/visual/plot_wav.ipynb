```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from scipy.fft import fft, fftfreq

samplerate, data = wavfile.read("filtered.wav")

if data.ndim == 1:
    # Mono-Datei
    signal = data
    mode = "mono"
elif data.ndim == 2 and data.shape[1] == 2:
    # Stereo-Datei
    signal = data.mean(axis=1)  # Stereo auf Mono mitteln
    mode = "stereo (gemittelt)"
else:
    raise ValueError("Unbekanntes Dateiformat")

n = len(signal)
xf = fftfreq(n, 1 / samplerate)[:n//2]
yf = np.abs(fft(signal))[:n//2]
yf_db = 20 * np.log10(yf / np.max(yf) + 1e-12)

plt.plot(xf, yf_db, lw=0.5)
plt.xlim(0, 22000)
plt.ylim(-120, 20)
plt.xlabel("Frequenz (Hz)")
plt.ylabel("Amplitude (dB)")
plt.grid(True)
plt.title(f"Frequenzgang: {mode}")
plt.show()
```
